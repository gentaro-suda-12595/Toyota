```sql
CREATE OR REPLACE FUNCTION solo_practice.silver.hr_region_filter(region_col STRING)
RETURN 
  CASE 
    -- Japanのみ表示
    WHEN current_user() = 'sudag@revolvesys.co.jp' AND region_col = 'Japan' THEN TRUE
    -- USのみ表示
    WHEN current_user() = 'nakahira@revolvesys.co.jp' AND region_col = 'US' THEN TRUE
    -- 管理者グループなどは全件表示（必要に応じて）
    WHEN is_account_group_member('admin_group') THEN TRUE
    -- それ以外は非表示
    ELSE FALSE
  END;


ALTER TABLE solo_practice.silver.silver_employee
SET ROW FILTER solo_practice.silver.hr_region_filter ON (region);
```
