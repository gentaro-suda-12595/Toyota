```sql
CREATE OR REPLACE FUNCTION solo_practice.silver.mask_salary(salary_col INT)
RETURN 
  CASE 
    -- 特権ユーザーとしてそのまま見せる
    WHEN current_user() = 'sudag@revolvesys.co.jp' THEN salary_col
    -- それ以外は0にする
    ELSE 0 
  END;

-- カラムへの適用
ALTER TABLE solo_practice.silver.silver_employee 
ALTER COLUMN salary SET MASK solo_practice.silver.mask_salary;
```
